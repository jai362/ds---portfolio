## Factors affecting Margin of Error
- Variability of Data (Standard Deviation)
- Sample Size
- Confidence level

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
```

## Relationship between Margin of Error and variability of data (Standard Deviation)

```python
sample_size = 30
confidence_level = 95

# calculate the critical value for the z-score
critical_value = stats.norm.ppf((1 + confidence_level/100) / 2)

# generate an array of population standard deviation
population_std = np.arange(1,101)

# calculate the margin or error for each population standard deviation 
# Margin of Error = critical_value × Standard Error
margin_of_errors = critical_value * (population_std / np.sqrt(sample_size))

# Plot the relationship 
plt.plot(population_std, margin_of_errors)
plt.xlabel("Population Standard Deviation")
plt.ylabel("Margin or Error")
plt.title("Relationship between Margin or Error and Population Standard Deviation")
plt.grid(True)
plt.show()
```

## Relationship between Margin of Error and Sample Size 

```python
population_std = 20
confidence_level = 95

# calculate the critical value for the z-score
critical_value = stats.norm.ppf((1 + confidence_level/100) / 2)

# generate an array of sample sizes
sample_sizes = np.arange(2,101)

# calculate the margin or error for each population standard deviation 
margin_of_errors = critical_value * (population_std / np.sqrt(sample_sizes))

# Plot the relationship
plt.plot(sample_sizes, margin_of_errors)
plt.xlabel("Sample Size")
plt.ylabel("Margin of Error")
plt.title("Relationship between Margin of Error and Sample Size")
plt.grid(True)
plt.show()
```

## Relationship between Margin of Error and Confidence level

```python
sample_size = 30
population_std = 20

# generate an array of confidence levels
confidence_levels = np.arange(50,100)

# calculate the critical value 
critical_values = stats.norm.ppf(((1 + confidence_levels)/100)/2)

# calculate the margin of error for each critical value
margin_of_errors = critical_values * (population_std / np.sqrt(sample_size))

# Plot the Relationship
plt.plot(confidence_levels, margin_of_errors)
plt.xlabel("Confidence levels")
plt.ylabel("Margin of Error")
plt.title("Relationship between Margin of Error and Confidence Levels")
plt.grid(True)
plt.show()
```

## Margin of Error (ME = z × σ / √n)* measures estimation uncertainty. Three key relationships:

### Sample Size Increases → Margin of Error Decreases
- Larger samples reduce ME because √n appears in the denominator ME shrinks with the square root of n, creating diminishing returns (4x more data needed to cut ME in half)

### Confidence Level Increases → Margin of Error Increases
- Higher confidence requires larger critical values (z*) Confidence and precision are inversely related; wider intervals are needed for greater certainty
- 
### Population Variability Increases → Margin of Error Increases
- Higher population standard deviation (σ) directly increases ME
