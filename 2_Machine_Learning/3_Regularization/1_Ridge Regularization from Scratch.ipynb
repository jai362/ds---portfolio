## Ridge Regression from Scratch on 2D Data 

```python
from sklearn.datasets import make_regression 

X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=20, random_state=13)

print(X.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import Ridge


X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=4)

R1 = Ridge(alpha=0.1, solver="cholesky")

R1.fit(X_train, y_train)

y_pred = R1.predict(X_test)

print("Slope Using Sklearn's Ridge: ",R1.coef_)
print("Intercept using Sklearn's Ridge: ",R1.intercept_,"\n")

print("R2 Score using Sklearn's Ridge: ",r2_score(y_test, y_pred))
```

## Now from Scratch

```python
class MeraRidgeND:
    def __init__(self, alpha=0.1):
        self.alpha = alpha
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        X_train = np.insert(X_train,0,1,axis=1)
        I = np.identity(X_train.shape[1])
        I[0][0] = 0
        result = np.linalg.inv(np.dot(X_train.T,X_train) + self.alpha*I).dot(X_train.T).dot(y_train)
        self.intercept_ = result[0]
        self.coef_ = result[1:]

        print("Slope Using My Own Ridge: ",self.coef_,"\n")
        print("Intercept using My Own Ridge: ",self.intercept_,"\n")

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
reg2 = MeraRidgeND()

reg2.fit(X_train, y_train)

y_pred3 = reg2.predict(X_test)

print("R2 Score using my Own Ridge: ",r2_score(y_test, y_pred3))
```

## Now Ridge Gradient Descent(Stochastic) from Scratch

```python
class MeraRidgeSGD:
    def __init__(self, epochs, learning_rate, alpha):
        self.coef_ = None
        self.intercept = None
        self.epochs = epochs
        self.learning_rate = learning_rate
        self.alpha = alpha

    def fit(self, X_train, y_train):
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        thetha = np.insert(self.coef_, 0, self.intercept_)  # insert value of self.intercept_ at 0 index in self.coef_ array

        X_train = np.insert(X_train,0,1,axis=1)

        for i in range(self.epochs):
            thetha_der = np.dot(X_train.T,X_train).dot(thetha) - np.dot(X_train.T,y_train) + self.alpha*thetha
            thetha = thetha - (self.learning_rate*thetha_der)

        self.coef_ = thetha[0]
        self.intercept_ = thetha[1:]

        print("Slope Using My Own Ridge: ",self.coef_,"\n")
        print("Intercept using My Own Ridge: ",self.intercept_,"\n")

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```
