## In this notebook We will learn Mini-Batch Gradient Descent from Scratch

```python
from sklearn.datasets import load_diabetes
import numpy as np
```

```python
X,y = load_diabetes(return_X_y = True)
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

## First Sklearn Implementation

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

lr = LinearRegression()
lr.fit(X_train, y_train)

print("Coefficient: ",lr.coef_)
print("\nIntercept: ",lr.intercept_)
```

```python
ypred = lr.predict(X_test)
print("R2 Score of Sklearn: ",r2_score(y_test, ypred))
```

## Now from Scratch

```python
class MBGDRegressor:
    def __init__(self, learning_rate, epochs, batch_size):
        self.intercept_ = None
        self.coef_ = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size

    def fit(self, X_train, y_train):
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        for i in range(self.epochs):
            for j in range(int(X_train.shape[0]/self.batch_size)):

                idx = np.random.choice(range(X_train.shape[0]),self.batch_size,replace=False)

                y_hat = np.dot(X_train[idx], self.coef_) + self.intercept_

                intercept_der = -2 * np.mean(y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr*intercept_der)

                coef_der = -2 * np.dot((y_train[idx] - y_hat),X_train[idx])
                self.coef_ = self.coef_ - (self.lr * coef_der)
        print("Coefficients: ",self.coef_)
        print("\nIntercept: ",self.intercept_)

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
MBGD = MBGDRegressor(0.01, 100,int(X_train.shape[0]/10))
```

```python
MBGD.fit(X_train, y_train)
```

```python
y_pred = MBGD.predict(X_test)
print("R2 Score of Mini-Batch Gradient Descent: ",r2_score(y_test, y_pred))
```

## Sklearn implementation of Mini-Batch Gradient Descent (with jugad)

```python
from sklearn.linear_model import SGDRegressor

SGD = SGDRegressor(learning_rate="constant", eta0=0.01)
```

```python
batch_size = 100
for epoch in range(100):
    indices = np.random.permutation(X_train.shape[0])

    for i in range(0, X_train.shape[0], batch_size):
        idx = indices[i:i+batch_size]

        SGD.partial_fit(X_train[idx], y_train[idx])

print("Coefficients: ",SGD.coef_)
print("\nIntercept: ",SGD.intercept_)
```

```python
y_pred1 = SGD.predict(X_test)
print("R2 Score: ",r2_score(y_test, y_pred1))
```
