## In this notebook We will learn Stochastic Gradient Descent from Scratch

```python
from sklearn.datasets import load_diabetes
import numpy as np
```

```python
X,y = load_diabetes(return_X_y=True)
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

## First Sklearn Implementation

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

lr = LinearRegression()
lr.fit(X_train, y_train)

print("Coefficients: ",lr.coef_)
print("\nIntercept: ",lr.intercept_)
```

```python
ypred = lr.predict(X_test)
print("R2 Score of Sklearn: ",r2_score(y_test, ypred))
```

## Now from Scratch

```python
class SGDRegressor:
    def __init__(self, learning_rate, epochs):
        self.intercept_ = None
        self.coef_ = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X_train, y_train):
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        for i in range(self.epochs):
            for j in range(X_train.shape[0]):
                ## Shuffle rows for this epoch (SGD processes each sample once, in random order)
                idx = np.random.permutation(len(X_train))
    
                y_hat = np.dot(X_train[idx], self.coef_) + self.intercept_
    
                coef_der = -2 * np.dot((y_train[idx] - y_hat),X_train[idx])
                self.coef_ = self.coef_ - (self.lr * coef_der)
    
                intercept_der = -2 * np.mean(y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr*intercept_der)

        print("Coefficients: ",self.coef_)
        print("\nIntercept: ",self.intercept_)

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
SGD = SGDRegressor(0.01, 20)
```

```python
SGD.fit(X_train, y_train)
```

```python
y_pred = SGD.predict(X_test)
print("R2 Score of Stochastic Gradient Descent: ",r2_score(y_test, y_pred))
```

## Stochastic Gradient Descent took only 20 epochs to match intercept of sklearn LinearRegression
- Batch works faster on fixed epochs but Stochastic works faster on bigger dataset
