## In this notebook We will do Two Case study of Chi-Square which is a Hypthesis test in statistics
- Goodness of fit test
- Test of Indendence

## Case Study - I (Goodness of fit test)

```python
import pandas as pd
from scipy.stats import chisquare
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
data = pd.read_csv(url)

data.head()
```

### we will use chi-square goodness of fit to see if the Pclass column is uniformly distributed

```python
# count passenger in each class 
class_counts = data["Pclass"].value_counts().sort_index()
class_counts
```

```python
# Calculated the expected counts assuming a uniform distribution
total_passengers = len(data)
expected_counts = total_passengers/3
expected = [expected_counts]*3
print("Expected passengers counts assuming (Uniform Distribution):\n", expected)
```

```python
# perform the chi-square goodness of fit test
chi2, p_value = chisquare(class_counts, expected)

print("Chi-Square Statistic: {:.2f}".format(chi2))
print("P-Value: {:.4f}".format(p_value))
```

```python
# interpret the result
alpha = 0.05
if p_value < alpha :
    print("We reject the null hypothesis. The distribution of passengers among the class is not uniform")
else:
    print("We failed to reject the null hypothesis")
```
