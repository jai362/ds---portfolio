### In this notebook we will prove how central limit theorem works
- Central limit theorem states that when we repeatedly take random samples from a population and calculate their means, those samples will form a normal distribution

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# set the parameters
num_population = 100000

# generate samples from uniform distribution
population_uniform = np.random.uniform(low=0, high=1, size=num_population)

plt.hist(population_uniform, bins=100, density=True, edgecolor="pink")
plt.title("Uniform Distribution")
plt.show()
```

```python
# now taking sample means 
num_samples = 1000
sample_size = 30
sample_means = []


for i in range(num_samples):
    samples_uniform = np.random.choice(population_uniform,sample_size)
    sample_mean = np.mean(samples_uniform)
    sample_means.append(sample_mean)

uniform_sample_means = np.array(sample_means)

# now plot the sample_means
sns.histplot(uniform_sample_means, kde=True, stat="density")
plt.title("Histplot of Sample Means")
plt.xlabel("Sample Means")
plt.ylabel("Density")
plt.show()
```

## Now test with pareto distribution

```python
num_population = 100000
a = 5

population_pareto = np.random.pareto(a, num_population)

plt.hist(population_pareto, bins=25, density=True, edgecolor="black")
plt.title("Pareto Distribution")
plt.show()
```

```python
num_samples = 1000
sample_size = 50
sample_means = []

for i in range(num_samples):
    sample = np.random.choice(population_pareto, sample_size)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)


pareto_sample_means = np.array(sample_means)

sns.histplot(pareto_sample_means, kde=True, stat="density")
plt.title("Histplot of Sample Means")
plt.xlabel("Sample means of Pareto distribution")
plt.ylabel("Density")
plt.show()
```

```python
# Conclusion/Summary
print("=" * 60)
print("CENTRAL LIMIT THEOREM DEMONSTRATION")
print("=" * 60)

print("\n Key Findings:\n")
print(f"1. UNIFORM DISTRIBUTION (Original):")
print(f"   - Shape: Flat/uniform")
print(f"   - Sample means converge to: {np.mean(uniform_sample_means):.4f}")

print(f"\n2. PARETO DISTRIBUTION (Original):")
print(f"   - Shape: Heavily skewed")
print(f"   - Sample means converge to: {np.mean(pareto_sample_means):.4f}")

print(f"\n3. ✨ MAGIC OF CLT:")
print(f"   - Despite different original distributions,")
print(f"   - Both sample means follow a NORMAL distribution!")
print(f"   - This proves the Central Limit Theorem works.")

print("\n📝 Conclusion:")
print("""
The Central Limit Theorem shows that:
✓ Original distribution shape doesn't matter
✓ Sample means always form a normal distribution
✓ This holds for uniform, pareto, exponential, etc.
✓ The larger the sample size, the better the result

This is why statistical tests (t-tests, confidence intervals)
work reliably even when data isn't normally distributed!
""")
```
