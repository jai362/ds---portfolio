```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Single sample t-test workflow for the titanic Age column

```python
train = pd.read_csv("5_train.csv").drop(columns=["Survived"])
test = pd.read_csv("5_test.csv")
```

```python
df = pd.concat([train, test]).sample(1309)
df.head()
```

```python
pop = df["Age"].dropna()
```

```python
sample_size = 25
num_sample = 20

samples = []
for i in range(num_sample):
    sample = pop.sample(n=sample_size, replace=False)  # replace=False Once an observation is selected, it cannot be selected again
    samples.append(sample)

samples = np.array(samples)
samples = samples.ravel()
samples.shape
```

```python
hypothesized_mean = 30
```

```python
import scipy.stats as stats

t_statistic, p_value = stats.ttest_1samp(samples, hypothesized_mean)
print("T-Statistic: ",t_statistic)
print("P-Value: ",p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

## Two Sample Independent T-test

## Case Study 2 : Average age of male is significantly greater than average age of female

```python
df.head()
```

```python
pop_male = df[df["Sex"] == "male"]["Age"].dropna()
pop_female = df[df["Sex"] == "female"]["Age"].dropna()
```

```python
sample_male = pop_male.sample(50)
sample_female = pop_female.sample(50)
```

### H0 -> Mean age of male and female are similar
### H1 -> Mean age of male is higher than Mean age of female

```python
t_statistic, p_value_two_tailed = stats.ttest_ind(sample_male, sample_female)

print("T-Statistic: ", t_statistic)
print("P-Value Two Tailed: ",p_value_two_tailed)
```

```python
if t_statistic > 0:  # Male mean > Female mean
    p_value_one_tailed = p_value_two_tailed / 2
else:
    p_value_one_tailed = 1 - (p_value_two_tailed / 2)
```

```python
alpha = 0.05

if p_value_one_tailed < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
