### Check normality of a distribution using QQ-Plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import scipy.stats as stats
```

```python
df = sns.load_dataset("iris")
df.head()
```

```python
sns.kdeplot(df["sepal_length"])
```

## Descriptive Statistics

```python
print(df["sepal_length"].skew())             # Close to zero Acceptable 
print(df["sepal_length"].kurtosis())         # acceptable range
```

### Check normality of a distribution using QQ-Plot

```python
fig = sm.qqplot(df["sepal_length"], line="45", fit=True)

plt.title("QQ Plot")
plt.xlabel("Theoretical  Quantiles")
plt.ylabel("Sample Quantiles")

plt.show()
```

### Points align closely with the diagonal line with only minor deviation at the tails so this indicates -> Approx Normality 

### Check Normality of distribution with Shapiro-Wilk test 

```python
stat, p_value = stats.shapiro(df["sepal_length"])

print(f"P-value-{p_value:.4f}")
print("Normal" if p_value > 0.05 else "Not Normal")
```

## SEPAL LENGTH NORMALITY ASSESSMENT
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

#### Visual Analysis (Q-Q Plot):
- Points align closely with the diagonal line with only minor deviations  at the tails → indicates APPROXIMATE NORMALITY

#### Statistical Test (Shapiro-Wilk):
- P-value = 0.0102 (< 0.05) → Technically rejects normality

#### Descriptive Statistics:
- Skewness = 0.314 (close to 0, acceptable)
- Kurtosis = -0.606 (acceptable range)

## FINAL CONCLUSION:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
### Data is APPROXIMATELY NORMAL

- Reasoning: Although the Shapiro-Wilk test is strict (p < 0.05), the Q-Q plot shows minimal deviations and skewness is acceptable (0.314). The data is suitable for parametric statistical methods and can be treated as approximately normal for practical purposes.
