## In this notebook we will learn about Regularization 
- Regularization is a penalty you add to the loss function to prevent the model from using large coefficients

```python
import numpy as np
import pandas as pd
```

## Ridge Regression from Scratch on 2D Data 

```python
from sklearn.datasets import make_regression 

X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=20, random_state=13)

print(X.shape)
```

## Baseline model to check the accuracy and coefficients from sklearn Ridge

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import Ridge


X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=4)

reg = Ridge(alpha=0.1, solver="cholesky")

reg.fit(X_train, y_train)

y_pred = reg.predict(X_test)

print("Accuracy of Sklearn's Ridge: ",r2_score(y_test, y_pred),"\n")

print("Slope Using Sklearn's Ridge: ",reg.coef_)
print("Intercept using Sklearn's Ridge: ",reg.intercept_)
```

## Now from Scratch for 2D data

```python
class MeraRidge:
    def __init__(self, alpha=0.1):
        self.alpha = alpha
        self.m = None
        self.b = None

    def fit(self, X_train, y_train):
        num = 0
        den = 0
        for i in range(X_train.shape[0]):
            num = num + (y_train[i] - y_train.mean())*(X_train[i] - X_train.mean())
            den = den + (X_train[i] - X_train.mean())**2

        self.m = num/(den + self.alpha)
        self.b = y_train.mean() - (self.m * X_train.mean())

        print("Slope Using My Own Ridge: ",reg.coef_)
        print("Intercept using My Own Ridge: ",reg.intercept_)

    def predict(self, X_test):
        return (self.m*X_test) + self.b
```

```python
MR = MeraRidge()
MR.fit(X_train, y_train)

y_pred1 = MR.predict(X_test)

print(r2_score(y_test, y_pred1))
```
